## Data Cleaning Decisions

- Dropped one corrupted row (Life Made WI-Fi Touchscreen Photo Frame) whose columns are shifted, so its Category is "1.9".
- Removed 483 exact duplicate rows, then removed repeated app names (698 more), keeping the row with the most reviews since it is the most up-to-date record.
- Reviews, Installs, Price and Size were stored as text. I converted them to numbers: Installs ("10,000+" becomes 10000), Price (removed the dollar sign), Size (converted to MB, with "Varies with device" left empty because it has no real value).
- Last Updated was converted to a real date.
- Type was rebuilt from Price (price above 0 means Paid), which also fixes the one missing value.
- I left the missing Rating values (about 1,460 apps) empty instead of filling them in. Filling them with the average would invent ratings and distort the rating distribution, so rating analyses just skip those apps.

```python
rows_before = len(apps)

# 1. corrupted row
apps = apps[apps['Category'] != '1.9']

# 2. exact duplicates
apps = apps.drop_duplicates()

# 3. repeated app names: keep the row with the most reviews
apps['Reviews'] = pd.to_numeric(apps['Reviews'])
apps = (apps.sort_values('Reviews', ascending=False)
            .drop_duplicates(subset='App', keep='first')
            .reset_index(drop=True))

print("Rows before:", rows_before, "| after:", len(apps))
```

```python
# fix data types
apps['Installs'] = apps['Installs'].str.replace('[+,]', '', regex=True).astype(int)
apps['Price'] = apps['Price'].str.replace('$', '', regex=False).astype(float)

def size_to_mb(s):
    if s == 'Varies with device':
        return np.nan
    if s.endswith('M'):
        return float(s[:-1])
    if s.endswith('k'):
        return float(s[:-1]) / 1024
    return np.nan

apps['Size_MB'] = apps['Size'].apply(size_to_mb)
apps['Last Updated'] = pd.to_datetime(apps['Last Updated'])
apps['Type'] = np.where(apps['Price'] > 0, 'Paid', 'Free')

print(apps.isnull().sum())
apps.dtypes
```

## Revenue Estimate

Estimated revenue per paid app = Price x Installs. This is only a rough estimate, because the Installs column is a lower bound ("10,000+" could mean any number above 10,000) and it ignores the store's cut and refunds.

I excluded the 20 paid apps priced above 100. They are joke "I am Rich" apps, and a handful of them would otherwise make categories like Lifestyle and Finance look far more profitable than they really are.


```python
paid['Revenue'] = paid['Price'] * paid['Installs']
paid_clean = paid[paid['Price'] <= 100]

revenue_by_cat = (paid_clean.groupby('Category')['Revenue'].sum()
                  .sort_values(ascending=False).head(10))

revenue_by_cat.sort_values().plot(kind='barh', figsize=(9, 6), color='seagreen')
plt.title('Estimated Revenue by Category (top 10, paid apps)')
plt.xlabel('Estimated Revenue')
plt.show()

paid_clean.sort_values('Revenue', ascending=False)[['App', 'Category', 'Price', 'Installs', 'Revenue']].head(5)
```

```python
# only reviews that actually have text can be analysed
rv = reviews.dropna(subset=['Translated_Review']).copy()
print("Usable reviews:", len(rv))

sia = SentimentIntensityAnalyzer()
rv['compound'] = rv['Translated_Review'].apply(lambda t: sia.polarity_scores(t)['compound'])

def label(c):
    if c >= 0.05:
        return 'Positive'
    elif c <= -0.05:
        return 'Negative'
    return 'Neutral'

rv['VADER'] = rv['compound'].apply(label)

rv['VADER'].value_counts().plot(kind='bar', color=['seagreen', 'firebrick', 'gray'], figsize=(7, 5))
plt.title('Review Sentiment (VADER)')
plt.xticks(rotation=0)
plt.ylabel('Number of Reviews')
plt.show()

rv['VADER'].value_counts(normalize=True) * 100
```

```python
print("Agreement with the file's own labels:", round((rv['VADER'] == rv['Sentiment']).mean() * 100, 1), "%")
pd.crosstab(rv['Sentiment'], rv['VADER'])
```

```python
merged = rv.merge(apps[['App', 'Category']], on='App', how='inner')
print("Reviews matched to apps:", len(merged), "| apps:", merged['App'].nunique())

cat_sent = merged.groupby('Category').agg(
    reviews=('compound', 'size'),
    avg_sentiment=('compound', 'mean'),
    pct_positive=('VADER', lambda s: (s == 'Positive').mean() * 100),
    pct_negative=('VADER', lambda s: (s == 'Negative').mean() * 100)
)

# ignore categories with too few reviews to be meaningful
cat_sent = cat_sent[cat_sent['reviews'] >= 100].sort_values('avg_sentiment')

cat_sent['avg_sentiment'].plot(kind='barh', figsize=(9, 10), color='steelblue')
plt.title('Average Review Sentiment by Category (VADER score)')
plt.xlabel('Average Sentiment Score')
plt.show()

cat_sent.sort_values('avg_sentiment', ascending=False).round(2)
```

## Conclusion: 3 Data-Driven Insights for a Developer Launching a New App

1. Avoid the most crowded categories unless the app is clearly different. Family (about 1,880 apps), Game (about 950) and Tools (about 830) have by far the most apps, while categories like Beauty, Comics and Parenting have only 50 to 60 each. The opportunity map shows that categories such as Communication, Video Players and Social have the highest average installs per app, so demand is there, but the competition and the big incumbents are too.

2. Plan for a free app. About 92% of apps are free, and only about 8% are paid. Paid apps do tend to be rated a little higher (about 4.26 against 4.17), but the pricing is low (median 2.99, and over 80% of paid apps cost 5 or less) and the revenue is concentrated in a few huge titles like Minecraft. A freemium model with in-app purchases looks safer than charging upfront.

3. Quality and user experience matter more than size, and users in some categories are much harder to please. App size has only a weak link with installs (rank correlation about 0.31), so there is no reason to compromise features to stay small. Review sentiment is highest in categories like Health and Fitness and Education, and lowest in Social, News and Video Players, where users are more critical. A new app in those categories needs to be clearly better than what exists, but it also means there is room to win users from disappointing alternatives.
