```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

apps = pd.read_csv('googleplaystore.csv')
reviews = pd.read_csv('googleplaystore_user_reviews.csv')

print("Apps:", apps.shape, "| Reviews:", reviews.shape)
apps.head()
```

```python
reviews.head()
```

```python
print(apps.isnull().sum())
print("Exact duplicate rows:", apps.duplicated().sum())
print("Repeated app names:", apps['App'].duplicated().sum())
apps.dtypes
```

```python
# the corrupted row with shifted columns
apps[apps['Category'] == '1.9']
```

## Data Cleaning Decisions

- Dropped one corrupted row (Life Made WI-Fi Touchscreen Photo Frame) whose columns are shifted, so its Category is "1.9".
- Removed 483 exact duplicate rows, then removed repeated app names (698 more), keeping the row with the most reviews since it is the most up-to-date record.
- Reviews, Installs, Price and Size were stored as text. I converted them to numbers: Installs ("10,000+" becomes 10000), Price (removed the dollar sign), Size (converted to MB, with "Varies with device" left empty because it has no real value).
- Last Updated was converted to a real date.
- Type was rebuilt from Price (price above 0 means Paid), which also fixes the one missing value.
- I left the missing Rating values (about 1,460 apps) empty instead of filling them in. Filling them with the average would invent ratings and distort the rating distribution, so rating analyses just skip those apps.

```python
rows_before = len(apps)

# 1. corrupted row
apps = apps[apps['Category'] != '1.9']

# 2. exact duplicates
apps = apps.drop_duplicates()

# 3. repeated app names: keep the row with the most reviews
apps['Reviews'] = pd.to_numeric(apps['Reviews'])
apps = (apps.sort_values('Reviews', ascending=False)
            .drop_duplicates(subset='App', keep='first')
            .reset_index(drop=True))

print("Rows before:", rows_before, "| after:", len(apps))
```

```python
# fix data types
apps['Installs'] = apps['Installs'].str.replace('[+,]', '', regex=True).astype(int)
apps['Price'] = apps['Price'].str.replace('$', '', regex=False).astype(float)

def size_to_mb(s):
    if s == 'Varies with device':
        return np.nan
    if s.endswith('M'):
        return float(s[:-1])
    if s.endswith('k'):
        return float(s[:-1]) / 1024
    return np.nan

apps['Size_MB'] = apps['Size'].apply(size_to_mb)
apps['Last Updated'] = pd.to_datetime(apps['Last Updated'])
apps['Type'] = np.where(apps['Price'] > 0, 'Paid', 'Free')

print(apps.isnull().sum())
apps.dtypes
```

```python
cat_counts = apps['Category'].value_counts()

plt.figure(figsize=(10, 10))
cat_counts.sort_values().plot(kind='barh', color='steelblue')
plt.title('Number of Apps per Category')
plt.xlabel('Number of Apps')
plt.show()

cat_counts.head(5)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 8))

apps['Rating'].dropna().plot(kind='hist', bins=30, color='teal', ax=axes[0])
axes[0].set_title('Distribution of App Ratings')
axes[0].set_xlabel('Rating')

cat_rating = apps.groupby('Category')['Rating'].mean().sort_values()
cat_rating.plot(kind='barh', color='orange', ax=axes[1])
axes[1].set_title('Average Rating by Category')
axes[1].set_xlim(3.5, 4.6)

plt.tight_layout()
plt.show()
```

```python
apps['Rating'].describe()
```

```python
sz = apps.dropna(subset=['Size_MB'])
sz_plot = sz[sz['Installs'] > 0]

plt.figure(figsize=(9, 6))
plt.scatter(sz_plot['Size_MB'], sz_plot['Installs'], alpha=0.3, color='purple')
plt.yscale('log')
plt.xlabel('App Size (MB)')
plt.ylabel('Installs (log scale)')
plt.title('App Size vs Number of Installs')
plt.show()

sz[['Size_MB', 'Installs']].corr(method='spearman')
```

```python
paid = apps[apps['Type'] == 'Paid'].copy()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
apps['Type'].value_counts().plot(kind='bar', color=['steelblue', 'firebrick'], ax=axes[0])
axes[0].set_title('Free vs Paid Apps')
axes[0].tick_params(axis='x', rotation=0)

paid[paid['Price'] <= 30]['Price'].plot(kind='hist', bins=30, color='firebrick', ax=axes[1])
axes[1].set_title('Price Distribution of Paid Apps (up to 30)')
axes[1].set_xlabel('Price')
plt.tight_layout()
plt.show()

print(apps['Type'].value_counts(normalize=True) * 100)
paid['Price'].describe()
```

```python
# a few paid apps cost almost 400: these are the joke "I am Rich" apps
print("Paid apps priced above 100:", (paid['Price'] > 100).sum())
paid[paid['Price'] > 100][['App', 'Price', 'Installs']].sort_values('Price', ascending=False).head(5)
```

## Revenue Estimate

Estimated revenue per paid app = Price x Installs. This is only a rough estimate, because the Installs column is a lower bound ("10,000+" could mean any number above 10,000) and it ignores the store's cut and refunds.

I excluded the 20 paid apps priced above 100. They are joke "I am Rich" apps, and a handful of them would otherwise make categories like Lifestyle and Finance look far more profitable than they really are.


```python
paid['Revenue'] = paid['Price'] * paid['Installs']
paid_clean = paid[paid['Price'] <= 100]

revenue_by_cat = (paid_clean.groupby('Category')['Revenue'].sum()
                  .sort_values(ascending=False).head(10))

revenue_by_cat.sort_values().plot(kind='barh', figsize=(9, 6), color='seagreen')
plt.title('Estimated Revenue by Category (top 10, paid apps)')
plt.xlabel('Estimated Revenue')
plt.show()

paid_clean.sort_values('Revenue', ascending=False)[['App', 'Category', 'Price', 'Installs', 'Revenue']].head(5)
```
